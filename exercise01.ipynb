### 🔐 Configuring API Access

Storing API keys directly in code is unsafe, so we’ll keep our secrets in environment variables managed by DataLab. Here’s the three‑step flow we’ll follow:

1️⃣ Generate a Tavily API key

1. Go to https://app.tavily.com/home.
2. Create your account.

3. Click “Create new API key.”

<img src="images/tavily-new-api-key.png" width="300">

4. Copy the key.  


2️⃣ Add the key as an environment variable in DataLab

1. Click the “Environment” tab in the notebook toolbar.  
2. Choose “Environment variables”, then click “Add”. 
3. Fill in the fields:  
   * Name: `TAVILY_API_KEY`  
   * Value: _paste the key from Step 1_

<img src="images/tavily-env-variable.png" width="500">

4. Click “Save”.


3️⃣ Connect the environment variable to this notebook session

1. In the “Environment variables” list, click the _three dots_ next to `TAVILY_API_KEY`.  
2. Choose “Connect”. 
3. Wait a few seconds for the kernel to restart (or restart manually via Run ▸ Restart kernel).  
   When the green dot appears, we’re good to go! ✅

```python
from openai import OpenAI as OpenAIClient

raw_client = OpenAIClient()

API_KEY   = raw_client.api_key   
API_BASE  = raw_client.base_url
```

With our API key in hand, and LlamaIndex installed, instantiating an LLM is one line:

```python
import os
from llama_index.llms.openai import OpenAI

API_BASE = "https://api.openai.com/v1"  # Define the API_BASE variable
API_KEY = os.environ["OPENAI_API_KEY"]
llm = OpenAI(model="gpt-4o-mini", api_key=API_KEY, api_base=API_BASE)
```

We can ask our LLM questions directly without an agent, just to check it's working:

```python
response = llm.complete("What's the capital of South Dakota?")
print(response.text)
```
